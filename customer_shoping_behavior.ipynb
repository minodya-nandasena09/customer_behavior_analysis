```python
import pandas as pd
```

```python
df = pd.read_csv('customer_shopping_behavior.csv')
```

```python
# read first 10 raws
df.head(10)
```

```python
#read numerical data
df.info()
```

```python
#read both numerical and categorical data
df.describe(include="all")
```

```python
# check for null values
df.isnull().sum()
```

```python
# filling the missing values of Review Rating using the median 
df['Review Rating'] = df.groupby('Category')['Review Rating'].transform(lambda x: x.fillna(x.median()))
```

```python
df.isnull().sum()
```

```python
# convert column headers into lowercase letters
df.columns = df.columns.str.lower()
# replace blanks with _
df.columns =  df.columns.str.replace(' ','_')
# remove (usd)
df=df.rename(columns={'purchase_amount_(usd)':'purchase_amount'})
```

```python
df.columns
```

```python
# create a column age_group and categorize ages into 4 major categories
# useful for marketings to understand cutomer behavoir
labels = ['Young Adult','Adult','Middle-Aged','Senior']
df['age_group'] = pd.qcut(df['age'], q=4, labels=labels)
```

```python
df[['age','age_group']].head(10)
```

```python
# create column purchace frquency to understand how often people do purchases
frequency_mapping = {
    'Fortnightly': 14,
    'Weekly': 7,
    'Monthly': 30,
    'Quarterly': 90,
    'Bi-weekly': 14,
    'Annually': 365,
    'Every 3 months': 90
}
df['purchase_frequency_days'] = df['frequency_of_purchases'].map(frequency_mapping)
```

```python
df[['purchase_frequency_days','frequency_of_purchases']].head(10)
```

```python
df[['discount_applied','promo_code_used']].head(10)
```
